```python
import numpy as np
from angerflow.layers.dense import Dense

# 1. Dummy Input banao:
# Maan lo 4 samples hain (batch size = 4), aur har sample ke 3 features hain (input_dim = 3)
X = np.random.randn(4, 3).astype(np.float32)

# 2. Dense layer banao:
# 5 output neurons chahiye
dense_layer = Dense(units=5, input_dim=3)

# 3. Forward pass run karo
output = dense_layer.forward(X)

print("Input shape: ", X.shape)       # (4, 3) aana chahiye
print("Output shape:", output.shape)  # (4, 5) aana chahiye
print("Weights shape:", dense_layer.W.shape) # (3, 5) aana chahiye
print("Bias shape:", dense_layer.b.shape)    # (1, 5) aana chahiye
print("Output:\n", output)
```
