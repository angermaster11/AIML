```python
# Similarity between numpy and pytorch
import numpy as np 
import torch 

a = np.array([2,3,4])
b = torch.tensor([2,3,4])

print(a*2)
print(b*2)
```

```python
# Conversion of Numpy and Pytorch

a_tensor = torch.from_numpy(a)
print(a_tensor)

a_back = a_tensor.numpy()
print(a_back)
print(type(a_back))
```

```python
# Creating Tensors with Multidimensional 

import torch 

a = torch.tensor(5) # Single Element
print(a)

b = torch.tensor([1,2,3,4,5]) # List of Element
print(b)

c = torch.tensor([[1,2,3],[4,5,6]]) # Grid of Element
print(c)
```

```python
# Check GPU and Cuda 

import torch 

print(torch.cuda.is_available())

if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))
    print(torch.cuda.device_count())
```

```python
# Using GPU in tensors 

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using ",device)

a = torch.randn(2,3)
print(a)
print(a.device)

b = a.to(device)
print(b.device)
```

```python
# Autograd 

import torch 

x = torch.tensor(2.0,requires_grad=True)
print(x)

y = x**2 

print("x", x.is_leaf)
print("x",x.grad_fn)

print("y",y.is_leaf)
print("y",y.grad_fn)

print(y.item())
print(y)

```

```python
temp = y.backward()
print(temp) # return None 
```
